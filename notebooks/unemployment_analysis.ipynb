```python
import sys
print(sys.executable)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
# ==========================================
# 2. LOADING THE DATASETS
# ==========================================

df = pd.read_csv("../data/Unemployment in India.csv")

df_2020 = pd.read_csv("../data/Unemployment_Rate_upto_11_2020.csv")

print("Both datasets loaded successfully!")
```

```python
# Display first five rows of Dataset 1

print("UNEMPLOYMENT IN INDIA")
display(df.head())
```

```python
# Display first five rows of Dataset 2

print("UNEMPLOYMENT RATE UP TO NOVEMBER 2020")
display(df_2020.head())
```

```python
print("Dataset 1:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nDataset 2:")
print("Rows:", df_2020.shape[0])
print("Columns:", df_2020.shape[1])
```

```python
print("Dataset 1 Columns:")
print(df.columns.tolist())

print("\nDataset 2 Columns:")
print(df_2020.columns.tolist())
```

```python
print("DATASET 1 INFORMATION")
df.info()

print("\nDATASET 2 INFORMATION")
df_2020.info()
```

```python
# ==========================================
# 3. MISSING VALUE ANALYSIS
# ==========================================

print("Missing Values - Dataset 1")
print(df.isnull().sum())

print("\nMissing Values - Dataset 2")
print(df_2020.isnull().sum())
```

```python
# Check completely empty rows

print("Completely empty rows in Dataset 1:",
      df.isnull().all(axis=1).sum())

print("Completely empty rows in Dataset 2:",
      df_2020.isnull().all(axis=1).sum())
```

```python
# ==========================================
# 4. DUPLICATE VALUE ANALYSIS
# ==========================================

print("Duplicate rows in Dataset 1:", df.duplicated().sum())

print("Duplicate rows in Dataset 2:", df_2020.duplicated().sum())
```

```python
print("Dataset 1 sample dates:")
print(df[" Date"].head(10))

print("\nDataset 2 sample dates:")
print(df_2020[" Date"].head(10))
```

```python
# ==========================================
# 5. CREATING CLEAN COPIES
# ==========================================

df_clean = df.copy()
df_2020_clean = df_2020.copy()

print("Clean copies created successfully!")
```

```python
# Remove rows where all values are missing

df_clean = df_clean.dropna(how="all").copy()

df_2020_clean = df_2020_clean.dropna(how="all").copy()

print("After removing completely empty rows:")

print("Dataset 1:", df_clean.shape)
print("Dataset 2:", df_2020_clean.shape)
```

```python
print("Dataset 1 duplicates:", df_clean.duplicated().sum())

print("Dataset 2 duplicates:", df_2020_clean.duplicated().sum())
```

```python
# Remove leading and trailing spaces from column names

df_clean.columns = df_clean.columns.str.strip()

df_2020_clean.columns = df_2020_clean.columns.str.strip()

print("Dataset 1 columns:")
print(df_clean.columns.tolist())

print("\nDataset 2 columns:")
print(df_2020_clean.columns.tolist())
```

```python
# Convert date columns

df_clean["Date"] = pd.to_datetime(
    df_clean["Date"].str.strip(),
    format="%d-%m-%Y",
    errors="coerce"
)

df_2020_clean["Date"] = pd.to_datetime(
    df_2020_clean["Date"].str.strip(),
    format="%d-%m-%Y",
    errors="coerce"
)

print("Date conversion completed!")
```

```python
print("DATASET 1 AFTER CLEANING")
df_clean.info()

print("\nDATASET 2 AFTER CLEANING")
df_2020_clean.info()
```

```python
print("Missing values in Dataset 1:")
print(df_clean.isnull().sum())

print("\nMissing values in Dataset 2:")
print(df_2020_clean.isnull().sum())
```

```python
# ==========================================
# 6. RESETTING DATASET INDEXES
# ==========================================

df_clean = df_clean.reset_index(drop=True)
df_2020_clean = df_2020_clean.reset_index(drop=True)

print("Dataset indexes reset successfully!")
```

```python
# ==========================================
# 7. DATE RANGE ANALYSIS
# ==========================================

print("DATASET 1 DATE RANGE")
print("Starting Date:", df_clean["Date"].min())
print("Ending Date:", df_clean["Date"].max())

print("\nDATASET 2 DATE RANGE")
print("Starting Date:", df_2020_clean["Date"].min())
print("Ending Date:", df_2020_clean["Date"].max())
```

```python
# ==========================================
# 8. DESCRIPTIVE STATISTICS
# ==========================================

print("DATASET 1 STATISTICS")

display(df_clean[
    [
        "Estimated Unemployment Rate (%)",
        "Estimated Employed",
        "Estimated Labour Participation Rate (%)"
    ]
].describe())
```

```python
print("DATASET 2 STATISTICS")

display(df_2020_clean[
    [
        "Estimated Unemployment Rate (%)",
        "Estimated Employed",
        "Estimated Labour Participation Rate (%)"
    ]
].describe())
```

```python
# ==========================================
# 9. REGIONAL ANALYSIS
# ==========================================

print("Dataset 1 - Number of unique regions:")
print(df_clean["Region"].nunique())

print("\nDataset 1 - Regions:")
print(sorted(df_clean["Region"].unique()))

print("\nDataset 2 - Number of unique regions:")
print(df_2020_clean["Region"].nunique())
```

```python
# ==========================================
# 10. AREA ANALYSIS
# ==========================================

print("Observations by Area:")

display(df_clean["Area"].value_counts())

print("\nAverage Unemployment Rate by Area:")

display(
    df_clean.groupby("Area")[
        "Estimated Unemployment Rate (%)"
    ].mean()
)
```

```python
# ==========================================
# 11. FREQUENCY ANALYSIS
# ==========================================

print("Dataset 1 Frequency:")
print(df_clean["Frequency"].value_counts())

print("\nDataset 2 Frequency:")
print(df_2020_clean["Frequency"].value_counts())
```

```python
# Remove extra spaces from frequency values
df_clean["Frequency"] = df_clean["Frequency"].str.strip()
df_2020_clean["Frequency"] = df_2020_clean["Frequency"].str.strip()

# Check the standardized values
print("Dataset 1 Frequency:")
print(df_clean["Frequency"].value_counts())

print("\nDataset 2 Frequency:")
print(df_2020_clean["Frequency"].value_counts())
```

```python
df_clean["Frequency"] = "Monthly"
df_2020_clean["Frequency"] = "Monthly"

print("Dataset 1:")
print(df_clean["Frequency"].value_counts())

print("\nDataset 2:")
print(df_2020_clean["Frequency"].value_counts())
```

```python
# Calculate average unemployment rate for each month
monthly_unemployment = (
    df_clean.groupby("Date")["Estimated Unemployment Rate (%)"]
    .mean()
    .reset_index()
)

# Display the monthly results
display(monthly_unemployment)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_unemployment["Date"],
    monthly_unemployment["Estimated Unemployment Rate (%)"],
    marker="o",
    linewidth=2
)

plt.title("Monthly Average Unemployment Rate in India (2019–2020)")
plt.xlabel("Month")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.savefig("../visualizations/monthly_unemployment_trend.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Calculate monthly average unemployment rate
monthly_unemployment_2020 = (
    df_2020_clean.groupby("Date")["Estimated Unemployment Rate (%)"]
    .mean()
    .reset_index()
)

display(monthly_unemployment_2020)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_unemployment_2020["Date"],
    monthly_unemployment_2020["Estimated Unemployment Rate (%)"],
    marker="o",
    linewidth=2
)

plt.title("Monthly Average Unemployment Rate in India (2020)")
plt.xlabel("Month")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.savefig(
    "../visualizations/monthly_unemployment_trend_2020.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Calculate average unemployment rate for each region
regional_unemployment = (
    df_2020_clean.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

display(regional_unemployment)
```

```python
# Select the 10 regions with the lowest unemployment rates
bottom_10_regions = regional_unemployment.tail(10).sort_values(
    by="Estimated Unemployment Rate (%)",
    ascending=True
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=bottom_10_regions,
    x="Estimated Unemployment Rate (%)",
    y="Region"
)

plt.title("10 Regions with Lowest Average Unemployment Rate (2020)")
plt.xlabel("Average Unemployment Rate (%)")
plt.ylabel("Region")
plt.tight_layout()

plt.savefig(
    "../visualizations/regional_unemployment_bottom10.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
top_10_regions = regional_unemployment.head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_10_regions,
    x="Estimated Unemployment Rate (%)",
    y="Region"
)

plt.title("Top 10 Regions with Highest Average Unemployment Rate (2020)")
plt.xlabel("Average Unemployment Rate (%)")
plt.ylabel("Region")
plt.tight_layout()

plt.savefig(
    "../visualizations/regional_unemployment_top10.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
regional_participation = (
    df_2020_clean.groupby("Region")[
        "Estimated Labour Participation Rate (%)"
    ]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

display(regional_participation)
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    data=regional_participation,
    x="Estimated Labour Participation Rate (%)",
    y="Region"
)

plt.title("Average Labour Participation Rate by Region (2020)")
plt.xlabel("Labour Participation Rate (%)")
plt.ylabel("Region")
plt.tight_layout()

plt.savefig(
    "../visualizations/labour_participation.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Calculate average unemployment and participation by region
regional_comparison = df_2020_clean.groupby("Region")[
    [
        "Estimated Unemployment Rate (%)",
        "Estimated Labour Participation Rate (%)"
    ]
].mean().reset_index()

# Create scatter plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=regional_comparison,
    x="Estimated Labour Participation Rate (%)",
    y="Estimated Unemployment Rate (%)",
    s=100
)

plt.title("Unemployment vs Labour Participation by Region (2020)")
plt.xlabel("Labour Participation Rate (%)")
plt.ylabel("Unemployment Rate (%)")
plt.tight_layout()

plt.savefig(
    "../visualizations/unemployment_vs_participation.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
correlation = regional_comparison[
    [
        "Estimated Unemployment Rate (%)",
        "Estimated Labour Participation Rate (%)"
    ]
].corr()

display(correlation)
```

```python
# Create a period column
df_clean["Period"] = np.where(
    df_clean["Date"] < "2020-03-01",
    "Before COVID-19",
    "During COVID-19"
)

# Calculate average unemployment by period
covid_comparison = (
    df_clean.groupby("Period")["Estimated Unemployment Rate (%)"]
    .mean()
    .reset_index()
)

display(covid_comparison)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=covid_comparison,
    x="Period",
    y="Estimated Unemployment Rate (%)"
)

plt.title("Unemployment Before and During COVID-19")
plt.xlabel("Period")
plt.ylabel("Average Unemployment Rate (%)")
plt.tight_layout()

plt.savefig(
    "../visualizations/covid_impact.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
before = covid_comparison.loc[
    covid_comparison["Period"] == "Before COVID-19",
    "Estimated Unemployment Rate (%)"
].values[0]

during = covid_comparison.loc[
    covid_comparison["Period"] == "During COVID-19",
    "Estimated Unemployment Rate (%)"
].values[0]

percentage_increase = ((during - before) / before) * 100

print(f"Before COVID-19: {before:.2f}%")
print(f"During COVID-19: {during:.2f}%")
print(f"Percentage increase: {percentage_increase:.2f}%")
```

```python
area_unemployment = (
    df_clean.groupby("Area")["Estimated Unemployment Rate (%)"]
    .mean()
    .reset_index()
    .sort_values(
        by="Estimated Unemployment Rate (%)",
        ascending=False
    )
)

display(area_unemployment)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=area_unemployment,
    x="Area",
    y="Estimated Unemployment Rate (%)"
)

plt.title("Average Unemployment Rate: Rural vs Urban")
plt.xlabel("Area")
plt.ylabel("Average Unemployment Rate (%)")
plt.tight_layout()

plt.savefig(
    "../visualizations/rural_urban_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Calculate unemployment by area and period
area_covid_comparison = (
    df_clean.groupby(["Area", "Period"])["Estimated Unemployment Rate (%)"]
    .mean()
    .reset_index()
)

display(area_covid_comparison)
```

```python
plt.figure(figsize=(9, 6))

sns.barplot(
    data=area_covid_comparison,
    x="Area",
    y="Estimated Unemployment Rate (%)",
    hue="Period"
)

plt.title("Rural vs Urban Unemployment Before and During COVID-19")
plt.xlabel("Area")
plt.ylabel("Average Unemployment Rate (%)")
plt.legend(title="Period")
plt.tight_layout()

plt.savefig(
    "../visualizations/rural_urban_covid_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Reshape the table for easier comparison
area_pivot = area_covid_comparison.pivot(
    index="Area",
    columns="Period",
    values="Estimated Unemployment Rate (%)"
)

# Calculate percentage increase
area_pivot["Percentage Increase (%)"] = (
    (area_pivot["During COVID-19"] - area_pivot["Before COVID-19"])
    / area_pivot["Before COVID-19"]
) * 100

display(area_pivot.round(2))
```

```python
correlation_matrix = df_2020_clean[
    [
        "Estimated Unemployment Rate (%)",
        "Estimated Employed",
        "Estimated Labour Participation Rate (%)"
    ]
].corr()

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Between Unemployment, Employment and Labour Participation")
plt.tight_layout()

plt.show()
```
